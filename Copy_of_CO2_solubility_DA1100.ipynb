```python
input_smiles = smiles['SMILES']
```

```python
from rdkit import Chem
from rdkit.Chem import AllChem
mol_list = []
for i in input_smiles:
  mol = Chem.MolFromSmiles(i)
  mol = Chem.AddHs(mol)  # Add hydrogens
  AllChem.EmbedMolecule(mol)  # Generate 3D coordinates
  AllChem.MMFFOptimizeMolecule(mol)  # Optimize geometry
  mol_list.append(mol)

threed_mol_list = mol_list

```

```python
from rdkit import Chem
from rdkit.Chem import AllChem, rdMolDescriptors

def get_molecular_volume(smiles):
  volume_list = []
  for i in smiles:
    mol = Chem.MolFromSmiles(i)
    mol = Chem.AddHs(mol)
    # Changed line: Create an EmbedParameters object and set randomSeed
    params = AllChem.ETKDG()
    params.randomSeed = 42
    AllChem.EmbedMolecule(mol, params)
    AllChem.UFFOptimizeMolecule(mol)
    # Estimate molecular volume (in Å³)
    vol = rdMolDescriptors.CalcExactMolWt(mol) / 1.4  # Rough estimate
    volume_list.append(vol)
  return volume_list

volume = get_molecular_volume(input_smiles)
```

```python
smiles['volume(in ang)'] = volume
```

```python
from rdkit import Chem

def count_halogens(smiles):
    halogens = {'F', 'Cl', 'Br', 'I'}
    count = []
    for i in smiles:
     mol = Chem.MolFromSmiles(i)
     count.append(sum(1 for atom in mol.GetAtoms() if atom.GetSymbol() in halogens))
    return count

def count_specific_atom(smiles, atom_symbol):
    count = []
    for i in smiles:
      mol = Chem.MolFromSmiles(i)
      count.append(sum(1 for atom in mol.GetAtoms() if atom.GetSymbol() == atom_symbol))
    return count

halogen_count = count_halogens(input_smiles)
O_count = count_specific_atom(input_smiles, 'O')
N_count = count_specific_atom(input_smiles, 'N')
S_count = count_specific_atom(input_smiles, 'S')


```

```python
smiles['halogen_count'] = halogen_count
smiles['O_count'] = O_count
smiles['N_count'] = N_count
smiles['S_count'] = S_count
```

```python
from rdkit import Chem
from rdkit.Chem import rdMolDescriptors

def get_tpsa(smiles):
  TPSA_list = []
  for i in smiles:
    mol = Chem.MolFromSmiles(i)
    TPSA_list.append(rdMolDescriptors.CalcTPSA(mol))
  return TPSA_list

TPSA = get_tpsa(input_smiles)

```

```python
smiles['TPSA(in ang)'] = TPSA
```

```python
smiles
```

```python
c_a_id = smiles['c_a_ID']
def match(value):
 k = {}
 for i in range(len(c_a_id)):
  k.update({c_a_id[i]:value[i]})
 return k

CA_vol = match(volume)
CA_halogen = match(halogen_count)
CA_O = match(O_count)
CA_N = match(N_count)
CA_S = match(S_count)
CA_TPSA = match(TPSA)
```

```python
C_test = test['c_id']
A_test = test['a_id']
def match_test(l_test,CA):
  k = []
  for i in l_test:
    if i in CA:
      k.append(CA[i])
    else:
      pass
  return k

halogen_count_c = match_test(C_test,CA_halogen)
halogen_count_a = match_test(A_test,CA_halogen)
O_count_c = match_test(C_test,CA_O)
O_count_a = match_test(A_test,CA_O)
N_count_c = match_test(C_test,CA_N)
N_count_a = match_test(A_test,CA_N)
S_count_c = match_test(C_test,CA_S)
S_count_a = match_test(A_test,CA_S)
volume_c = match_test(C_test,CA_vol)
volume_a = match_test(A_test,CA_vol)
TPSA_c = match_test(C_test,CA_TPSA)
TPSA_a = match_test(A_test,CA_TPSA)

def add(l1,l2):
  k = []
  for i in range(len(l1)):
    k.append(l1[i]+l2[i])
  return k

tot_halogen_count = add(halogen_count_c,halogen_count_a)
tot_O_count = add(O_count_c,O_count_a)
tot_N_count = add(N_count_c,N_count_a)
tot_S_count = add(S_count_c,S_count_a)
tot_volume = add(volume_c,volume_a)
tot_TPSA = add(TPSA_c,TPSA_a)

```

```python
test['halogen_count'] = tot_halogen_count
test['O_count'] = tot_O_count
test['N_count'] = tot_N_count
test['S_count'] = tot_S_count
test['volume(in ang)'] = tot_volume
test['TPSA(in ang)'] = tot_TPSA
```
